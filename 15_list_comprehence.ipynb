# LIST COMPREHENSION
# List comprehension is a short way to create a new list from an existing list or range


```python
numbers = [1, 2, 3, 4, 5]
# We create an empty list and then add each number.
result = []
for x in numbers:
    result.append(x)

print("Normal:", result)
```

```python
# x value that will be stored
# for x take each value from numbers
result = [x for x in numbers]

print("Comprehension:", result)
```

```python
#SQUARE OF NUMBERS
numbers = [1, 2, 3, 4, 5]
squares = [x**2 for x in numbers]
print("Squares:", squares)
```

```python
#  DOUBLE EACH NUMBER
double = [x * 2 for x in numbers]
print("Double:", double)
```

```python
# GET ONLY EVEN NUMBERS
numbers = [1, 2, 3, 4, 5, 6]
even = [x for x in numbers if x % 2 == 0]
print("Even:", even)
```

```python
# GET ONLY ODD NUMBERS
odd = [x for x in numbers if x % 2 != 0]
print("Odd:", odd)
```

```python
# NUMBERS GREATER THAN 3
numbers = [1, 2, 3, 4, 5, 6]
greater = [x for x in numbers if x > 3]
print("Greater than 3:", greater)
```

```python
# SQUARE ONLY EVEN NUMBERS
numbers = [1, 2, 3, 4, 5, 6]
even_squares = [x**2 for x in numbers if x % 2 == 0]
print("Square of even numbers:", even_squares)
```

```python
# IF-ELSE IN LIST COMPREHENSION
numbers = [1, 2, 3, 4, 5, 6]
result = [x if x % 2 == 0 else x * 2 for x in numbers]
print("Result:", result)
```

```python
# WORKING WITH STRINGS
names = ["ayush", "rahul", "rohan"]
upper_names = [name.upper() for name in names]
print("Uppercase:", upper_names)
```

```python
#  FILTER STRINGS
names = ["ram", "ayush", "rahul", "raj"]
long_names = [name for name in names if len(name) > 4]
print("Long names:", long_names)
```

```python
# GET LENGTH OF EACH NAME
lengths = [len(name) for name in names]
print("Lengths:", lengths)
```

```python
# RANGE WITH LIST COMPREHENSION
squares = [x ** 2 for x in range(1, 11)]
print("1 to 10 Squares:", squares)
```

```python
# MULTIPLES OF 5
multiples = [x for x in range(1, 51) if x % 5 == 0]
print("Multiples of 5:", multiples)
```

```python
# POSITIVE NUMBERS
numbers = [-5, -2, 0, 3, 7, -1, 10]
positive = [x for x in numbers if x > 0]
print("Positive:", positive)
```

```python
# NEGATIVE NUMBERS
negative = [x for x in numbers if x < 0]
print("Negative:", negative)
```

```python
# FLATTEN A 2D LIST
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
flat = [value for row in matrix for value in row]

print("Flattened:", flat)
```

```python
# EVEN NUMBERS FROM 2D LIST
even = [
    value
    for row in matrix
    for value in row
    if value % 2 == 0
]

print("Even values:", even)
```

```python
# USER INPUT
data = input("Enter numbers separated by space: ")
numbers = [int(x) for x in data.split()]
print("Your numbers:", numbers)
```

```python
# SQUARE USER INPUT
squares = [x ** 2 for x in numbers]
print("Squares:", squares)
```

```python
# FILTER EVEN USER INPUT
even = [x for x in numbers if x % 2 == 0]
print("Even numbers:", even)
```
