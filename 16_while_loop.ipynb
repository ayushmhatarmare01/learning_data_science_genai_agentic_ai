```python
# REVERSE NUMBER
num = int(input("Enter number: "))
reverse = 0
while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num //= 10
print("Reverse =", reverse)
```

```python
# PALINDROME
# A palindrome is the same forward and backward.
num = int(input("Enter number: "))
original = num
reverse = 0
while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num //= 10
if original == reverse:
    print("Palindrome")
else:
    print("Not palindrome")
```

```python
# Check whether a number is prime
num = int(input("Enter number: "))
if num < 2:
    print("Not Prime")
else:
    i = 2
    while i < num:
        if num % i == 0:
            print("Not Prime")
            break
        i += 1
    else:
        print("Prime")
```

```python
# PERFECT NUMBER
num = int(input("Enter number: "))
i = 1
total = 0
while i < num:
    if num % i == 0:
        total += i
    i += 1
if total == num:
    print("Perfect Number")
else:
    print("Not a Perfect Number")
```

```python
# ARMSTRONG NUMBER
num = int(input("Enter number: "))
original = num
digits = len(str(num))
total = 0
while num > 0:
    digit = num % 10
    total += digit ** digits
    num //= 10
if total == original:
    print("Armstrong Number")
else:
    print("Not Armstrong Number")
```

```python
# Fibonacci series
n = int(input("Enter number of terms: "))
a = 0
b = 1
i = 0
while i < n:
    print(a, end=" ")
    a, b = b, a + b
    i += 1
print()
```

```python
# GCD using Euclidean Algorithm
# GCD = Greatest Common Divisor
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
while b != 0:
    a, b = b, a % b
print("GCD =", a)
```

```python
# LCM = Least Common Multiple
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
if a > b:
    lcm = a
else:
    lcm = b
while True:
    if lcm % a == 0 and lcm % b == 0:
        break
    lcm += 1
print("LCM =", lcm)
```

```python
# Find both GCD and LCM
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
original_a = a
original_b = b
while b != 0:
    a, b = b, a % b
gcd = a
lcm = (original_a * original_b) // gcd
print("GCD =", gcd)
print("LCM =", lcm)
```

```python
# DECIMAL TO BINARY
num = int(input("Enter decimal number: "))
if num == 0:
    print("Binary = 0")
else:
    binary = ""
    while num > 0:
        remainder = num % 2
        binary = str(remainder) + binary
        num //= 2
    print("Binary =", binary)
```

```python
# Binary to Decimal
binary = int(input("Enter binary number: "))
decimal = 0
power = 0
while binary > 0:
    digit = binary % 10
    decimal += digit * (2 ** power)
    binary //= 10
    power += 1
print("Decimal =", decimal)
```

```python
# Find largest digit
num = int(input("Enter number: "))
largest = 0
while num > 0:
    digit = num % 10
    if digit > largest:
        largest = digit
    num //= 10
print("Largest digit =", largest)
```

```python
# Find smallest digit
num = int(input("Enter number: "))
smallest = 9
while num > 0:
    digit = num % 10
    if digit < smallest:
        smallest = digit
    num //= 10
print("Smallest digit =", smallest)
```

```python
# Check whether a number contains  0
num = int(input("Enter number: "))

found = False

while num > 0:

    digit = num % 10

    if digit == 0:
        found = True
        break

    num //= 10

if found:
    print("Number contains 0")
else:
    print("Number does not contain 0")
```
