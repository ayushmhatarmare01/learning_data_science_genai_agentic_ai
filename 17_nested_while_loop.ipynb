```python
# Nested loop means:
# One while loop is inside another while loop.
# Outer loop  = usually controls ROWS
# Inner loop  = usually controls COLUMNS
```

```python
row = 1

while row <= 3:

    col = 1

    while col <= 3:
        print("*", end=" ")
        col += 1

    print()
    row += 1
```

```python
#  MULTIPLICATION TABLES
# Outer loop = table number
# Inner loop = multiplier

i = 1

while i <= 5:

    j = 1

    while j <= 10:
        print(i, "x", j, "=", i * j)
        j += 1

    print()
    i += 1
```

```python
# MULTIPLICATION TABLE OF USER'S NUMBER
# Only one table is printed.
# Inner while loop performs the multiplication.

num = int(input("Enter number: "))

i = 1

while i <= 10:

    print(num, "x", i, "=", num * i)

    i += 1
```

```python
#  PRINT ROW-COLUMN GRID
# Print the position of every element.

row = 1

while row <= 3:

    col = 1

    while col <= 3:

        print(row, col, end=" | ")

        col += 1

    print()
    row += 1
```

```python
# NESTED LOOP WITH IF
# Print only even numbers in a 3 x 5 grid.

row = 1

while row <= 3:

    col = 1

    while col <= 5:

        if col % 2 == 0:
            print(col, end=" ")

        col += 1

    print()
    row += 1
```

```python
#  PRINT MULTIPLES USING NESTED LOOP
num = 1
while num <= 5:
    i = 1

    while i <= 5:

        print(num * i, end=" ")

        i += 1

    print()
    num += 1
```

```python
# 18. RIGHT-ALIGNED STAR TRIANGLE
row = 1
while row <= 5:
    space = 1
    while space <= 5 - row:
        print(" ", end=" ")
        space += 1
    star = 1
    while star <= row:
        print("*", end=" ")
        star += 1
    print()
    row += 1
```

```python
#  PYRAMID PATTERN
# Spaces decrease while stars increase.
row = 1

while row <= 5:

    space = 1

    while space <= 5 - row:
        print(" ", end=" ")
        space += 1

    star = 1

    while star <= (2 * row - 1):
        print("*", end=" ")
        star += 1

    print()
    row += 1
```

```python
#  INVERTED PYRAMID
# Start with many stars and reduce them.
row = 5

while row >= 1:

    space = 1

    while space <= 5 - row:
        print(" ", end=" ")
        space += 1

    star = 1

    while star <= (2 * row - 1):
        print("*", end=" ")
        star += 1

    print()
    row -= 1
```

```python
#  NUMBER PYRAMID
row = 1
while row <= 5:
    space = 1
    while space <= 5 - row:
        print(" ", end=" ")
        space += 1
    col = 1
    while col <= row:
        print(row, end=" ")
        col += 1
    print()
    row += 1
```

```python
# NESTED WHILE WITH USER INPUT
rows = int(input("Enter rows: "))
columns = int(input("Enter columns: "))

row = 1

while row <= rows:

    col = 1

    while col <= columns:

        print("*", end=" ")

        col += 1

    print()
    row += 1
```

```python
#  PRINT 1 TO N IN ROWs
n = int(input("Enter n: "))
row = 1
number = 1

while number <= n:

    col = 1

    while col <= 5 and number <= n:

        print(number, end=" ")

        number += 1
        col += 1

    print()
    row += 1
```

```python
#  NESTED WHILE WITH BREAK
# break stops the INNER loop.

row = 1

while row <= 5:

    col = 1

    while col <= 5:

        if col == 4:
            break

        print(col, end=" ")

        col += 1

    print()
    row += 1
```

```python
# NESTED WHILE WITH CONTINUE
# Skip even numbers and print odd numbers.

row = 1

while row <= 3:

    col = 0

    while col < 5:

        col += 1

        if col % 2 == 0:
            continue

        print(col, end=" ")

    print()
    row += 1
```

```python

#  NESTED LOOP WITH SUM
# Find the sum of numbers in each row.

row = 1

while row <= 3:

    col = 1
    total = 0

    while col <= 5:

        total += col

        col += 1

    print("Row", row, "Sum =", total)

    row += 1
```

```python
# NESTED LOOP FOR MATRIX
# A matrix has rows and columns.
# Nested loops are perfect for processing it.

matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

row = 0

while row < len(matrix):

    col = 0

    while col < len(matrix[row]):

        print(matrix[row][col], end=" ")

        col += 1

    print()
    row += 1
```

```python
# MATRIX SUM
# Add every element of a matrix.

matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

row = 0
total = 0

while row < len(matrix):

    col = 0

    while col < len(matrix[row]):

        total += matrix[row][col]

        col += 1

    row += 1

print("Matrix Sum =", total)
```

```python
#  NESTED LOOP - MULTIPLICATION GRID
# Creates a multiplication grid from 1 to 5.

row = 1

while row <= 5:

    col = 1

    while col <= 5:

        print(row * col, end="\t")

        col += 1

    print()
    row += 1
```

```python
#  MENU INSIDE WHILE LOOP
# # Outer while keeps the menu running.
# Inner logic handles the selected operation.

while True:

    print()
    print("1. Print Hello")
    print("2. Print Python")
    print("3. Exit")

    choice = int(input("Enter choice: "))

    if choice == 1:

        print("Hello!")

    elif choice == 2:

        print("Python!")

    elif choice == 3:

        print("Program ended.")
        break

    else:

        print("Invalid choice")
```

```python
#  DIAMOND PATTERN
# Upper half
row = 1

while row <= 5:

    space = 1

    while space <= 5 - row:
        print(" ", end=" ")
        space += 1

    star = 1

    while star <= 2 * row - 1:
        print("*", end=" ")
        star += 1

    print()
    row += 1


# Lower half
row = 4

while row >= 1:

    space = 1

    while space <= 5 - row:
        print(" ", end=" ")
        space += 1

    star = 1

    while star <= 2 * row - 1:
        print("*", end=" ")
        star += 1

    print()
    row -= 1
```

```python
# HOLLOW RIGHT TRIANGLE
n = 5
row = 1
while row <= n:

    col = 1

    while col <= row:

        if col == 1 or col == row or row == n:
            print("*", end=" ")
        else:
            print(" ", end=" ")

        col += 1

    print()
    row += 1
```

```python
# 27. HOLLOW SQUARE
n = 5
row = 1
while row <= n:
    col = 1
    while col <= n:

        if row == 1 or row == n or col == 1 or col == n:
            print("*", end=" ")
        else:
            print(" ", end=" ")

        col += 1

    print()
    row += 1
```
